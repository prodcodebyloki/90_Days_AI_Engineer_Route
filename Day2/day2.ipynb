```python
from dotenv import load_dotenv
from agents import Agent, Runner, trace
#Runner and Trace are concepts (with helper classes/functions) used to execute and observe AI workflows—not core Python features.
```

```python
load_dotenv(override=True)
```

```python
agent=Agent(name="Jokestar",
instructions="you are joke teller which can tell jokes related to openaAI",
 model='gpt-4o-mini')
```

```python
agent 
```

```python
result=await Runner.run(agent,input="tell me a joke")
print(result.final_output)
```
